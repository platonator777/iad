# Сделки купли-продажи • 2020–2021

1. Считаем завершённые сделки по месяцам и выбираем активный период.
2. Строим карту «товар × цена» и делим товары на три ценовых сегмента.
3. Сравниваем средние цены сегментов t-тестом Уэлча на более поздних сделках.
4. Ищем товар с наибольшей доходностью на вложения и смотрим, как он показал себя в следующем месяце.

Источник — дневные CSV из папки `traders_sells` рядом с ноутбуком. Для повторного запуска нужны pandas, NumPy, SciPy, Matplotlib и Jupyter. Графики сохраняются в папку `traders_sells_outputs`.

```python
from itertools import combinations
from pathlib import Path

import matplotlib.pyplot as plt
from matplotlib.colors import LogNorm
from matplotlib.ticker import FuncFormatter, NullLocator, PercentFormatter
import numpy as np
import pandas as pd
from scipy.stats import ttest_ind
from IPython.display import display

SOURCE = Path("traders_sells")
OUTPUT = Path("traders_sells_outputs")
OUTPUT.mkdir(exist_ok=True)

BLUE = "#17658F"
TEAL = "#008074"
ORANGE = "#A75116"
GRAY = "#64748B"
INK = "#182638"
GRID = "#DFE6EC"
plt.rcParams.update({
    "font.family": "DejaVu Sans",
    "font.size": 10,
    "axes.titleweight": "bold",
    "axes.titlepad": 12,
    "axes.spines.top": False,
    "axes.spines.right": False,
    "axes.edgecolor": GRID,
    "axes.labelcolor": INK,
    "text.color": INK,
    "xtick.color": GRAY,
    "ytick.color": GRAY,
    "grid.color": GRID,
    "axes.axisbelow": True,
    "figure.dpi": 105,
})

def number(value, digits=0):
    return f"{value:,.{digits}f}".replace(",", " ").replace(".", ",")

def show_figure(fig, filename):
    fig.savefig(OUTPUT / f"{filename}.png", dpi=175, bbox_inches="tight", facecolor="white")
    plt.show()
    plt.close(fig)
```

## 1. Когда сделок было больше всего?

Одна строка описывает завершённую покупку и последующую продажу. Товар определяет поле `name_id`. Месяц сделки определяем по `sold_time`. Удаляем совпадения по всем шести содержательным полям. `trader_sold` встречается повторно в разных строках, поэтому дедупликация по нему одному убрала бы и различающиеся записи. Валюта в файлах не указана. Цены и прибыль приводим в единицах датасета.

```python
columns = ["trader_sold", "name_id", "sold_time", "bought_time", "bought_price", "sold_price"]
files = sorted(SOURCE.glob("*.csv"))
if not files:
    raise FileNotFoundError(f"CSV не найдены в {SOURCE.resolve()}")

raw = pd.concat(
    (pd.read_csv(path, usecols=columns).assign(file_date=path.stem) for path in files),
    ignore_index=True,
)
sales = raw.drop_duplicates(subset=columns).copy()
sales["sold_time"] = pd.to_datetime(sales["sold_time"])
sales["bought_time"] = pd.to_datetime(sales["bought_time"])
prices = sales[["bought_price", "sold_price"]]
assert sales[columns].notna().all().all(), "Есть пропуски в полях сделок"
assert (np.isfinite(prices) & prices.gt(0)).all().all(), "Есть некорректные цены"
assert sales["sold_time"].ge(sales["bought_time"]).all(), "Продажа раньше покупки"

sales["sold_date"] = sales["sold_time"].dt.normalize()
sales["month"] = sales["sold_time"].dt.to_period("M").astype(str)
monthly = sales.groupby("month").agg(
    deals=("name_id", "size"), active_days=("sold_date", "nunique"), items=("name_id", "nunique")
)
monthly["per_day"] = monthly["deals"] / monthly["active_days"]
monthly["change_pct"] = monthly["deals"].pct_change().mul(100)
print(f"{len(files)} файлов, {number(len(raw))} строк, удалено полных дублей: {len(raw) - len(sales)}.")
display(monthly.rename(columns={
    "deals": "Сделки", "active_days": "Дни с продажами", "items": "Товары",
    "per_day": "Сделок за активный день", "change_pct": "К прошлому месяцу, %",
}).round(1))

active_months = monthly.index[monthly.index.to_series().between("2020-01", "2020-05")]
fig, ax = plt.subplots(figsize=(11.4, 4.1), layout="constrained")
positions = np.arange(len(monthly))
bars = ax.bar(positions, monthly["deals"], width=0.72,
              color=[BLUE if month in active_months else "#A8B8C9" for month in monthly.index])
ax.bar_label(bars, labels=[number(n) for n in monthly["deals"]], padding=4, fontsize=8)
ax.axvline(len(active_months) - 0.5, color=GRAY, linestyle="--", linewidth=1)
ax.set(xticks=positions, xticklabels=monthly.index, ylabel="Сделки",
       title="Число завершённых сделок по месяцам",
       ylim=(0, monthly["deals"].max() * 1.16))
ax.tick_params(axis="x", labelrotation=40)
ax.yaxis.set_major_formatter(FuncFormatter(lambda x, _: number(x)))
ax.grid(axis="y")
show_figure(fig, "01_monthly_deals")

active = sales[sales["month"].isin(active_months)].copy()
active["profit"] = active["sold_price"] - active["bought_price"]
active["holding_days"] = (active["sold_time"] - active["bought_time"]).dt.total_seconds() / 86_400
active["is_loss"] = active["profit"].lt(0)
print(f"В выбранном периоде {number(len(active))} сделок.")
```

**Выбираем январь–май 2020 года.** В апреле наблюдается пик. После него май возвращается примерно к мартовскому уровню, а в июне сделок уже на 35% меньше, чем в мае. В последующие месяцы объём остаётся ниже майского. Среднее за активный день подтверждает просадку. Последний месяц выгрузки представлен по 27 марта 2021 года.

## 2. Редкие товары, необычные цены и тепловая карта

Редкие товары и необычные цены требуют разных проверок. Если у товара мало сделок, его медиана может зависеть от одной или двух продаж. Сначала посмотрим, сколько товаров и сделок остаётся при разных порогах, используя январь–февраль.

```python
formation = active[active["month"].lt("2020-03")]
all_items = (formation.groupby("name_id")["sold_price"]
             .agg(deals="size", median_price="median")
             .sort_values(["median_price", "name_id"]))
min_group_deals = 50
thresholds = [1, 10, 20, 30, min_group_deals, 100]
coverage = []
for threshold in thresholds:
    kept = all_items.loc[all_items["deals"].ge(threshold)]
    boundaries = kept["median_price"].quantile([1/3, 2/3])
    coverage.append({
        "Минимум сделок": threshold,
        "Товары": len(kept),
        "Сделки": kept["deals"].sum(),
        "Доля сделок, %": 100 * kept["deals"].sum() / len(formation),
        "Нижний тертиль": boundaries.iloc[0],
        "Верхний тертиль": boundaries.iloc[1],
    })
coverage = pd.DataFrame(coverage).set_index("Минимум сделок")
display(coverage.round(2))

activity = all_items["deals"].value_counts().sort_index(ascending=False)
remaining = activity.cumsum().sort_index()
covered = (activity * activity.index).cumsum().sort_index() / len(formation)
fig, axes = plt.subplots(1, 2, figsize=(10.7, 3.8), layout="constrained", sharex=True)
for ax, curve, title in zip(
    axes, [remaining, covered.mul(100)],
    ["Сколько товаров остаётся", "Какая доля сделок остаётся"],
):
    ax.step(curve.index, curve, where="pre", color=BLUE, linewidth=2)
    ax.axvline(min_group_deals, color=ORANGE, linestyle="--", linewidth=1.1)
    value = curve.loc[min_group_deals]
    ax.scatter(min_group_deals, value, color=ORANGE, zorder=3)
    label = number(value) if ax is axes[0] else f"{number(value, 1)}%"
    ax.annotate(label, (min_group_deals, value), xytext=(9, 8),
                textcoords="offset points", color=INK)
    ax.set(xscale="log", xlim=(1, 250), xlabel="Минимум сделок на товар",
           title=title, ylim=(0, None))
    ax.set_xticks([1, 5, 10, 20, 50, 100, 200], labels=["1", "5", "10", "20", "50", "100", "200"])
    ax.xaxis.set_minor_locator(NullLocator())
    ax.grid(axis="y")
axes[0].yaxis.set_major_formatter(FuncFormatter(lambda x, _: number(x)))
axes[1].yaxis.set_major_formatter(PercentFormatter(100, decimals=0))
show_figure(fig, "06_trade_coverage")
```

### Как читается карта

Выбрали порог **50 сделок суммарно за январь–февраль**. Он оставляет 952 из 5 428 товаров и 79,7% сделок за эти два месяца. С порогом 20 в анализ вошли бы 1 666 товаров и 90,6% сделок, а с порогом 100 — 555 товаров и 66,1% сделок. Это рабочий порог для читаемой карты и более устойчивых медиан. Тертильные границы в таблице показывают, что состав групп зависит от порога.

Чтобы редкая цена не растягивала цветовую карту, отмечаем крайние продажи **внутри каждого товара**. На логарифмах цен вычисляем квартильные границы и отступаем на три межквартильных размаха с каждой стороны. При нулевом размахе фильтр для товара пропускаем. Такие записи исключаем только из карты и отдельной проверки устойчивости t-теста. Само по себе попадание за границу не доказывает ошибку, а цена дорогого товара может быть обычной для него.

Строка карты — один ID товара, столбец — ценовой интервал на логарифмической оси. Чем темнее ячейка, тем больше сделок. Сортируем товары по исходной медиане цены продажи. Правая панель показывает 30 товаров, равномерно выбранных по порядковым позициям. Для трёх сопоставимых по размеру групп используем тертили медианных цен и округляем их до 6 и 25. Горизонтальные линии отделяют группы, вертикальные показывают пороги.

```python
items = all_items.loc[all_items["deals"].ge(min_group_deals)].copy()
tertiles = items["median_price"].quantile([1/3, 2/3])
low, high = tertiles.round().astype(int)
groups = [f"До {low}", f"От {low} до {high}", f"От {high}"]
group_colors = [BLUE, TEAL, ORANGE]
items["group"] = pd.cut(items["median_price"], [0, low, high, np.inf],
                        right=False, labels=groups)
print(f"Тертили: {number(tertiles.iloc[0], 2)} и {number(tertiles.iloc[1], 2)}.")
display(items.groupby("group", observed=True).agg(
    Товары=("deals", "size"), Сделки=("deals", "sum"),
    Минимальная_медиана=("median_price", "min"),
    Максимальная_медиана=("median_price", "max"),
).round(2))

eligible_formation = formation[formation["name_id"].isin(items.index)]
log_prices = np.log(eligible_formation["sold_price"])
quartiles = log_prices.groupby(eligible_formation["name_id"]).quantile([.25, .75]).unstack()
iqr = quartiles[.75] - quartiles[.25]
fences = pd.DataFrame({"lower": quartiles[.25] - 3 * iqr,
                       "upper": quartiles[.75] + 3 * iqr})

def price_outliers(data):
    log_price = np.log(data["sold_price"])
    lower = data["name_id"].map(fences["lower"])
    upper = data["name_id"].map(fences["upper"])
    return lower.lt(upper) & (log_price.lt(lower) | log_price.gt(upper))

excluded_on_map = price_outliers(eligible_formation)
heat_sales = eligible_formation.loc[~excluded_on_map]
print(f"Для карты убрано {number(excluded_on_map.sum())} крайних цен "
      f"из {number(len(eligible_formation))} сделок.")
price_edges = np.geomspace(heat_sales["sold_price"].min(),
                           heat_sales["sold_price"].max(), 61)
item_rows = items.index.get_indexer(heat_sales["name_id"])
frequency, _, _ = np.histogram2d(item_rows, heat_sales["sold_price"],
                                 bins=[np.arange(len(items) + 1), price_edges])
assert int(frequency.sum()) == len(heat_sales), "Часть сделок выпала из тепловой карты"
sample_rows = np.linspace(0, len(items) - 1, 30).round().astype(int)
cmap = plt.colormaps["Blues"].copy()
cmap.set_bad("white")
norm = LogNorm(vmin=1, vmax=frequency.max())

fig, axes = plt.subplots(1, 2, figsize=(12, 7.1), layout="constrained",
                         gridspec_kw={"width_ratios": [1.24, 1]})
for ax, rows, title in zip(axes, [np.arange(len(items)), sample_rows],
                           [f"Все {len(items)} товаров", "30 товаров крупным планом"]):
    grid = ax.pcolormesh(price_edges, np.arange(len(rows) + 1),
                         np.ma.masked_equal(frequency[rows], 0),
                         cmap=cmap, norm=norm, shading="flat")
    ax.scatter(items["median_price"].iloc[rows], np.arange(len(rows)) + 0.5,
               s=6 if len(rows) == len(items) else 13, marker="|", color=ORANGE,
               linewidths=0.9, label="Медиана товара")
    boundaries = np.flatnonzero(items["group"].iloc[rows].to_numpy()[1:] !=
                                 items["group"].iloc[rows].to_numpy()[:-1]) + 1
    for boundary in boundaries:
        ax.axhline(boundary, color=GRAY, linewidth=0.8, linestyle=":")
    for cutoff in (low, high):
        ax.axvline(cutoff, color=GRAY, linewidth=0.8, linestyle=":")
    labels = np.linspace(0, len(rows) - 1, 14).round().astype(int) if len(rows) > 30 else np.arange(len(rows))
    price_ticks = sorted({2, low, high, 100, 500, 2_000, 8_000})
    price_ticks = [tick for tick in price_ticks if tick <= price_edges[-1]]
    ax.set(xscale="log", ylim=(len(rows), 0), yticks=labels + 0.5,
           yticklabels=items.index[rows[labels]].astype(str), title=title,
           xlabel="Цена продажи, ед. (логарифмическая шкала)", ylabel="ID товара")
    ax.set_xticks(price_ticks, labels=[number(tick) for tick in price_ticks])
    ax.xaxis.set_minor_locator(NullLocator())
    ax.tick_params(axis="y", labelsize=8, length=0)
    ax.legend(loc="upper right", frameon=False, fontsize=9)
fig.suptitle("Частота продаж по товарам и ценовым интервалам • январь–февраль",
             fontsize=14, fontweight="bold")
fig.colorbar(grid, ax=axes, label="Сделок в ячейке (логарифмическая шкала)", shrink=0.87)
show_figure(fig, "02_price_heatmap")
```

## 3. Различаются ли средние цены сегментов?

Фиксируем группы по январю–февралю. Затем для каждого товара считаем его **среднюю цену продажи в марте–мае**. Так единица наблюдения в тесте — товар, и товар с сотней продаж имеет тот же вес, что товар с десятью. Товары без продаж в проверочном периоде исключаются. Разброс цен неодинаков, поэтому сравниваем пары двусторонним t-тестом Уэлча и корректируем три p-значения методом Холма при уровне значимости 0,05. Нулевая гипотеза для каждой пары — равенство средних цен.

Дополнительно повторяем проверку на марте–апреле с теми же группами, чтобы проверить чувствительность к включению мая.

Основной тест использует все проверочные продажи отобранных товаров. Отдельно повторяем тест после удаления крайних цен по границам января–февраля. При этом сохраняем первоначальные группы.

```python
def holdout_prices(end_month, trim_prices=False):
    period = active.loc[active["month"].between("2020-03", end_month)
                        & active["name_id"].isin(items.index)]
    if trim_prices:
        period = period.loc[~price_outliers(period)]
    return (period
            .groupby("name_id")["sold_price"].agg(test_deals="size", mean_price="mean")
            .join(items[["group"]], how="inner"))

def welch_comparisons(data, labels):
    results = []
    for cheaper, dearer in combinations(labels, 2):
        a = data.loc[data["group"].eq(cheaper), "mean_price"].to_numpy()
        b = data.loc[data["group"].eq(dearer), "mean_price"].to_numpy()
        test = ttest_ind(b, a, equal_var=False, alternative="two-sided")
        interval = test.confidence_interval(confidence_level=0.95)
        results.append({
            "pair": f"«{dearer}» − «{cheaper}»",
            "difference": b.mean() - a.mean(),
            "ci_low": interval.low,
            "ci_high": interval.high,
            "t": test.statistic,
            "p": test.pvalue,
        })
    result = pd.DataFrame(results).sort_values("p")
    multipliers = np.arange(len(result), 0, -1)
    result["p_holm"] = np.minimum(1, np.maximum.accumulate(result["p"].to_numpy() * multipliers))
    return result.sort_index()

checked = holdout_prices("2020-05")
summary = checked.groupby("group", observed=True).agg(
    items=("mean_price", "size"), deals=("test_deals", "sum"),
    mean=("mean_price", "mean"), sd=("mean_price", "std"),
)
print(f"В проверку вошёл {len(checked)} из {len(items)} товаров.")
display(summary.rename(columns={
    "items": "Товары", "deals": "Сделки", "mean": "Средняя цена",
    "sd": "Стандартное отклонение",
}).round(2))
tests = welch_comparisons(checked, groups)
tests_without_may = welch_comparisons(holdout_prices("2020-04"), groups)
checked_trimmed = holdout_prices("2020-05", trim_prices=True)
tests_trimmed = welch_comparisons(checked_trimmed, groups)
trimmed_summary = checked_trimmed.groupby("group", observed=True)["mean_price"].agg(["size", "mean"])
removed = checked["test_deals"].sum() - checked_trimmed["test_deals"].sum()
print(f"Из дополнительной проверки исключено {number(removed)} "
      f"из {number(checked['test_deals'].sum())} сделок с крайними ценами.")
display(trimmed_summary.rename(columns={
    "size": "Товары после отсечения", "mean": "Средняя цена после отсечения",
}).round(2))
report = tests.assign(p_holm_mar_apr=tests_without_may["p_holm"],
                      p_holm_trimmed=tests_trimmed["p_holm"]).rename(columns={
    "pair": "Пара", "difference": "Разница", "ci_low": "95% ДИ, нижняя",
    "ci_high": "95% ДИ, верхняя", "t": "t", "p": "p",
    "p_holm": "p Холма", "p_holm_mar_apr": "p Холма без мая",
    "p_holm_trimmed": "p Холма без крайних цен",
})
p_columns = ["p", "p Холма", "p Холма без мая", "p Холма без крайних цен"]
display(report.style.hide(axis="index").format({**{c: "{:.2e}" for c in p_columns},
    **{c: "{:.2f}" for c in ["Разница", "95% ДИ, нижняя", "95% ДИ, верхняя", "t"]}}))

fig, axes = plt.subplots(1, 2, figsize=(11.8, 4.2), layout="constrained",
                         gridspec_kw={"width_ratios": [1, 1.25]})
ax = axes[0]
values = [checked.loc[checked["group"].eq(group), "mean_price"] for group in groups]
boxes = ax.boxplot(values, patch_artist=True, showmeans=True,
                   meanprops={"marker": "D", "markerfacecolor": INK, "markeredgecolor": "white"},
                   medianprops={"color": INK, "linewidth": 1.4},
                   flierprops={"markersize": 2.5, "alpha": 0.4})
for box, color in zip(boxes["boxes"], group_colors):
    box.set_facecolor(color)
    box.set_edgecolor(color)
    box.set_alpha(0.48)
ax.set(yscale="log", title="Распределение средних цен товаров",
       ylabel="Средняя цена товара, ед. (лог. шкала)")
ax.set_xticks([1, 2, 3], labels=[f"{group}\nn = {n}" for group, n in zip(groups, summary["items"])])
ax.grid(axis="y")
ax.plot([], [], "D", color=INK, markersize=5, label="Среднее сегмента")
ax.legend(frameon=False, loc="upper left", fontsize=8)

ax = axes[1]
ax.axvline(0, linestyle="--", color=GRAY, linewidth=1)
ax.errorbar(tests["difference"], np.arange(len(tests)),
            xerr=np.array([tests["difference"] - tests["ci_low"],
                           tests["ci_high"] - tests["difference"]]),
            fmt="o", color=BLUE, capsize=4)
ax.set(yticks=np.arange(len(tests)), yticklabels=tests["pair"],
       xlabel="Разница средних цен, ед.", title="Разница средних и 95% интервалы Уэлча")
ax.invert_yaxis()
ax.grid(axis="x")
show_figure(fig, "03_group_comparisons")
```

**Все три пары различаются после поправки Холма.** Результат сохраняется при проверке только на марте–апреле и при исключении крайних цен. После отсечения в марте–мае осталось 949 товаров, а средние составили 4,05, 13,28 и 181,61. На графике показаны обычные 95% интервалы для отдельных пар. Оценки относятся к активно торговавшимся товарам из этой выгрузки. Разделение по цене в начале периода и проверка позже показывают устойчивость ценовой сегментации. Общие рыночные шоки могут связывать цены разных товаров, тогда стандартное предположение теста о независимости наблюдений выполняется лишь приблизительно.

### Почему три группы?

При двух группах пропадает отдельный средний сегмент. Четыре и пять групп дают более детальную разбивку, но с менее запоминающимися границами. Три группы дают два простых порога и примерно по 300 товаров в каждой. Это выбор для интерпретации графика. Тепловая карта показывает плавный переход, поэтому число естественных кластеров по ней определить нельзя.

Для сравнения разбиваем те же товары на 2–5 групп по точным квантилям января–февраля и считаем тесты на марте–мае. Поправка Холма в таблице относится к парам внутри каждого варианта. Эти варианты нужны для оценки чувствительности, основной вариант из трёх групп с округлёнными порогами остаётся выше.

```python
alternatives = []
for k in [2, 3, 4, 5]:
    cuts = items["median_price"].quantile(np.arange(1, k) / k)
    labels = [f"Группа {i}" for i in range(1, k + 1)]
    assigned = pd.cut(items["median_price"], [0, *cuts, np.inf], right=False, labels=labels)
    alternative = checked.assign(group=assigned.reindex(checked.index))
    sizes = alternative.groupby("group", observed=True).size()
    comparisons = welch_comparisons(alternative, labels)
    alternatives.append({
        "Группы": k,
        "Пороги, ед.": " / ".join(number(cut, 2) for cut in cuts),
        "Товаров в группах": " / ".join(map(str, sizes)),
        "Значимые пары": f"{comparisons['p_holm'].lt(.05).sum()} из {len(comparisons)}",
        "Макс. p Холма": f"{comparisons['p_holm'].max():.2e}",
    })
display(pd.DataFrame(alternatives).set_index("Группы"))
```

## 4. Какой товар выгоднее перепродавать?

**Основной критерий — валовая доходность на вложения (ROI).** Для товара складываем разницу между ценой продажи и ценой покупки по завершённым сделкам и делим на суммарную стоимость их покупки. Сохраняем все корректные сделки, включая редкие цены. Рейтинг строим по январю–апрелю и смотрим, что произошло с теми же товарами в мае.

Показываем две версии рейтинга. **Все товары** отвечают на вопрос о максимуме в исторических данных, включая разовые сделки. **Регулярно торговавшиеся товары** с минимум 30 завершёнными сделками в каждом месяце января–апреля позволяют оценить, насколько результат повторялся. Для обоих вариантов также показываем максимум по прибыли на сделку и по суммарной валовой прибыли.

```python
def trade_metrics(data):
    result = data.groupby("name_id").agg(
        deals=("profit", "size"), cost=("bought_price", "sum"),
        gross_profit=("profit", "sum"), profit_per_deal=("profit", "mean"),
        median_buy=("bought_price", "median"), median_sell=("sold_price", "median"),
        median_days=("holding_days", "median"), loss_share=("is_loss", "mean"),
    )
    return result.assign(roi=result["gross_profit"] / result["cost"])

ranking_period = active[active["month"].lt("2020-05")]
may = active[active["month"].eq("2020-05")]
min_activity = (ranking_period.groupby(["name_id", "month"]).size().unstack(fill_value=0)
                .reindex(columns=active_months[:-1], fill_value=0).min(axis=1))
training_metrics = trade_metrics(ranking_period).join(min_activity.rename("min_monthly"))
eligible = training_metrics.query("min_monthly >= 30").sort_values(
    ["roi", "deals"], ascending=False, kind="stable"
)
may_metrics = trade_metrics(may)
winner = eligible.index[0]
leaders = eligible.head(10).join(may_metrics[["roi", "deals"]].add_prefix("may_"))
print(f"Кандидатов: {len(eligible)}. Лидер по ROI: {winner}.")
print("Лидер при порогах 20, 30 и 50 сделок в месяц:",
      [int(training_metrics.loc[min_activity.ge(n), "roi"].idxmax()) for n in [20, 30, 50]])

metrics_ru = {
    "deals": "Сделки", "cost": "Вложения", "gross_profit": "Валовая прибыль",
    "profit_per_deal": "Прибыль на сделку", "median_buy": "Медиана покупки",
    "median_sell": "Медиана продажи", "median_days": "Медиана дней до продажи",
    "loss_share": "Доля убыточных, %", "roi": "ROI, %",
    "may_roi": "ROI в мае, %", "may_deals": "Сделки в мае",
}

def show_metrics(table, fields):
    table = table[fields].copy()
    for column in ("roi", "loss_share", "may_roi"):
        if column in table:
            table[column] *= 100
    display(table.rename(columns=metrics_ru).round(2))

show_metrics(leaders, ["deals", "cost", "profit_per_deal", "roi", "may_deals", "may_roi"])
criteria_rows = []
criteria_names = {
    "roi": "Доходность на вложения",
    "profit_per_deal": "Прибыль на сделку",
    "gross_profit": "Суммарная валовая прибыль",
}
for scope, table in [("Все товары", training_metrics), ("Регулярные продажи", eligible)]:
    for metric, criterion in criteria_names.items():
        item_id = table[metric].idxmax()
        row = table.loc[item_id]
        may_row = may_metrics.reindex([item_id]).iloc[0]
        criteria_rows.append({
            "Выборка": scope,
            "Критерий": criterion,
            "ID товара": item_id,
            "Сделки в январе–апреле": int(row["deals"]),
            "Валовая прибыль на сделку": number(row["profit_per_deal"], 2),
            "Валовая прибыль всего": number(row["gross_profit"], 2),
            "ROI, %": number(row["roi"] * 100, 2),
            "Сделки в мае": int(may_row["deals"]) if pd.notna(may_row["deals"]) else 0,
        })
display(pd.DataFrame(criteria_rows).set_index(["Выборка", "Критерий"]))

fig, ax = plt.subplots(figsize=(10.4, 4.8), layout="constrained")
y = np.arange(len(leaders))
ax.hlines(y, leaders["roi"] * 100, leaders["may_roi"] * 100, color="#BAC7D2", linewidth=1.6)
ax.scatter(leaders["roi"] * 100, y, s=39, color=BLUE, label="Январь–апрель", zorder=3)
ax.scatter(leaders["may_roi"] * 100, y, s=36, color=ORANGE, marker="D", label="Май", zorder=3)
ax.set(yticks=y, yticklabels=leaders.index.astype(str), xlabel="Валовой ROI на вложения",
       ylabel="ID товара", title="Топ-10 по ROI среди регулярно торговавшихся товаров")
ax.invert_yaxis()
ax.xaxis.set_major_formatter(PercentFormatter(100, decimals=0))
ax.set_xlim(left=0)
ax.grid(axis="x")
ax.legend(loc="lower right", frameon=False)
show_figure(fig, "04_roi_leaders")
```

```python
winner_deals = active.loc[active["name_id"].eq(winner)]
winner_monthly = winner_deals.groupby("month").agg(
    deals=("profit", "size"), cost=("bought_price", "sum"), gross_profit=("profit", "sum"),
    median_buy=("bought_price", "median"), median_sell=("sold_price", "median"),
    median_days=("holding_days", "median"), loss_share=("is_loss", "mean"),
)
winner_monthly["roi"] = winner_monthly["gross_profit"] / winner_monthly["cost"]
show_metrics(winner_monthly, ["deals", "median_buy", "median_sell",
                             "gross_profit", "roi", "median_days", "loss_share"])
winner_periods = pd.DataFrame({
    "Январь–апрель": eligible.loc[winner],
    "Май": may_metrics.loc[winner],
}).T
show_metrics(winner_periods, ["deals", "median_buy", "median_sell", "median_days",
                             "cost", "gross_profit", "profit_per_deal", "roi", "loss_share"])

fig, (prices_ax, deals_ax) = plt.subplots(2, 1, figsize=(9.4, 5.9), sharex=True,
                                        layout="constrained", height_ratios=[1.65, 1])
x = np.arange(len(winner_monthly))
prices_ax.plot(x, winner_monthly["median_sell"], color=BLUE, marker="o", label="Продажа")
prices_ax.plot(x, winner_monthly["median_buy"], color=TEAL, marker="s", label="Покупка")
prices_ax.set(ylabel="Медианная цена, ед.", ylim=(0, winner_monthly["median_sell"].max() * 1.26))
prices_ax.legend(frameon=False, loc="center right")
prices_ax.grid(axis="y")

bars = deals_ax.bar(x, winner_monthly["deals"], width=0.48,
                    color=[BLUE] * (len(x) - 1) + [ORANGE])
deals_ax.bar_label(bars, padding=3)
deals_ax.set(ylabel="Сделки", ylim=(0, winner_monthly["deals"].max() * 1.22),
             xticks=x, xticklabels=winner_monthly.index)
deals_ax.grid(axis="y")
for ax in (prices_ax, deals_ax):
    ax.axvline(x[-1] - 0.5, linestyle="--", linewidth=1, color=GRAY)
fig.suptitle(f"Товар {winner} • цены и число завершённых сделок",
             fontsize=14, fontweight="bold")
show_figure(fig, "05_winner_history")

floor = sales["sold_price"].min()
print(f"Доля продаж лидера по минимальной цене датасета ({number(floor, 2)}): "
      f"{number(winner_deals['sold_price'].eq(floor).mean() * 100, 1)}%.")
```
